## 6. Merging, Joining, Concatenating

#### Objectives
- `pd.merge` for relational joins
- `concat` for stacking
- `join` as shorthand

---
Relational joins combine rows based on keys, much like SQL.

```python
left = pd.DataFrame({'id':[1,2], 'name':['Ana','Ben']})
right = pd.DataFrame({'id':[1,2], 'sales':[250, 100]})
merged = pd.merge(left, right, on='id')
merged
```

## 9. Time Series

#### Objectives
- `DatetimeIndex`
- Resampling
- Rolling windows

---
Pandas makes time‑based indexing first‑class.

```python
rng = pd.date_range('2025-01-01', periods=6, freq='D')
ts = pd.Series(np.random.randn(len(rng)), index=rng)
weekly = ts.resample('W').mean()
weekly
```

## 11. Performance Optimization

#### Objectives
- Vectorization vs loops
- `eval` & `query`
- Memory efficiency

---
Prefer vectorized operations; use `pd.eval` for large expressions.

```python
n = 10_000
big = pd.DataFrame({'a': np.random.rand(n), 'b': np.random.rand(n)})
%timeit big['c'] = big['a'] + big['b']
```
